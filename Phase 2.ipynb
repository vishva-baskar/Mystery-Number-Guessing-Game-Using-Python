### 2.2 while loop - when end number is not defined and the instruction has to execute infinite times

```python
secret_number = 74

while True:
    user_number = int(input('Enter your gussing number :'))
    
    if secret_number == user_number:
        print('Congragulation your gussing is correct!')
        break
    else:
        print('Oops your gussing is wrong')
```
