### 4.2 Eliminate number of attempts by using while loop

```python
import random

print('Welcome to Number Guessing Game !!!')
print('Guess the Number between 1 to 100')

secret_number = random.randint(1,100)
attempts = 0
while True: 

    user_input =int(input('Enter the Number'))
    attempts += 1

    if user_input < 1 or user_input >100 :
        print('Invali Number enter the number between 1 to 100')
    elif user_input > secret_number :
        print(user_input, 'is grater than the secret number')
    elif user_input < secret_number :
         print(user_input, 'is lesser than the secert number')
    else:
        print('You guessed the correct number')
        print(f'Congragulation you guessed the secret number {secret_number} in {attempts} attemptes ')
        break
```
